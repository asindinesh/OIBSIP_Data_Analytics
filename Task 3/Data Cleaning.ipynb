# Missing Value Handling

```python
df.isnull().sum()
```

```python
print(df.columns)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df = df.dropna(subset=["Cabin"])
```

# Remove Duplicate Rows

```python
duplicates_before = df.duplicated().sum()

print("Duplicates before:", duplicates_before)

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicates after:", duplicates_after)
```

# Standardization

```python
df["Sex"] = df["Sex"].str.strip().str.title()
```

```python
df["Embarked"] = df["Embarked"].str.strip().str.upper()
```

```python
df["Name"] = df["Name"].str.strip()
```

# Detect Outliers

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
outliers = df[(df["Age"] < lower) | (df["Age"] > upper)]

print(outliers)
```

```python
df["Age"] = np.where(df["Age"] < lower, lower, df["Age"])
df["Age"] = np.where(df["Age"] > upper, upper, df["Age"])
```

# Data Type Correction

```python
df.dtypes
```

```python
df["PassengerId"] = df["PassengerId"].astype(str)
```

# Before vs After Summary

```python
rows_before = len(df)
null_before = df.isnull().sum().sum()
duplicates_before = df.duplicated().sum()
```

```python
rows_after = len(df)
null_after = df.isnull().sum().sum()
duplicates_after = df.duplicated().sum()
```

```python
summary = pd.DataFrame({
    "Metric": ["Rows", "Null Values", "Duplicate Rows"],
    "Before": [rows_before, null_before, duplicates_before],
    "After": [rows_after, null_after, duplicates_after]
})

summary
```
